### Transaction Type Distribution.

```python
df["type"].value_counts()
```

**Initial Observation**

CASH_OUT is the most frequent transaction type in the dataset.

However, transaction volume alone does not determine fraud risk.
Further analysis is needed to compare fraud counts and fraud rates
across transaction types.

```python
df['type'].value_counts().tail(1)
```

**Key Findings**
- CASH_OUT is the most frequent transaction type.
- DEBIT is the least frequent transaction type, with 41,432 transactions.
- Transaction volume alone is not sufficient to determine fraud risk.

```python
df["isFraud"].value_counts()
```

### Fraud Distribution

**Key Findings:**
- Legitimate transactions: 6,354,407
- Fraudulent transactions: 8,213
- Fraud represents approximately 0.129% of all transactions.

**Interpretation:**
The dataset is highly imbalanced, with fraudulent transactions
representing a very small proportion of the total.

This imbalance should be considered when evaluating fraud
detection strategies.

```python
fraud_rate= df["isFraud"].mean()*100
```

```python
print(fraud_rate)
```

```python
df.groupby("type")["isFraud"].sum()
```

### Fraud Count by Transaction Type

**Key Findings:**
- CASH_OUT recorded 4,116 fraudulent transactions.
- TRANSFER recorded 4,097 fraudulent transactions.
- No fraud was recorded for CASH_IN, DEBIT, or PAYMENT.
- All 8,213 fraudulent transactions occurred in CASH_OUT and TRANSFER.

```python
fraud_rate_by_type = df.groupby("type")["isFraud"].mean() * 100

print(fraud_rate_by_type)
```

### Fraud Rate by Transaction Type

**Key Findings:**
- TRANSFER has the highest fraud rate at approximately 0.769%.
- CASH_OUT has a fraud rate of approximately 0.184%.
- CASH_OUT records slightly more fraudulent transactions, but TRANSFER has a substantially higher fraud rate.
- No fraudulent transactions were recorded in CASH_IN, DEBIT, or PAYMENT.

**Business Insight:**
TRANSFER and CASH_OUT transactions deserve closer attention in fraud monitoring. However, transaction volume alone does not determine fraud risk.
